Probabilidade do time vencer o jogo a partir do desempenho nos últimos jogos:

<br>

*Probabilidade do time A vencer:*

$$P(a) = \frac{S_a}{S_a + S_b}$$

*Probabilidade do time B vencer:*

$$P(b) = \frac{S_b}{S_a + S_b}$$

# Brasil vs. Japão

```python
import numpy as np
```

```python
#  Cartão V., Cartão A. Gols sof. (CG)
Brasil_cg = np.array([0, 5, 1])
Japao_cg = np.array([0, 1, 3])
```

```python
# Pesos Cartões e Gols sofridos:
p_cg = np.array([1000, 500, 1000])
```

```python
Brasil_cg_pd = Brasil_cg * p_cg
```

```python
Japao_cg_pd = Japao_cg * p_cg
```

```python
# Desempenho 3 jogos Brasil:
Brasil = np.array([7, 2, 1680, 1512, 43, 
                   14, 6, 40, 18, 29, 
                   11, 40, 16, 43, 0, 
                   0, 26, 24, 21, 57, 
                   42, 499, 360, 290, 
                   486, 392, 1168, 10, 
                   108, 737])
```

```python
# Desempeho 3 jogos Japão:
Japao = np.array([7, 1, 1403, 1218, 50, 
                  14, 7, 27, 9, 14, 
                  13, 42, 11, 30, 0, 
                  0, 16, 13, 20, 49, 
                  64, 446, 282, 328, 
                  424, 420, 1172, 4, 105, 
                  756])
```

```python
# Pesos Desempenho em jogo:
p_d = np.array([500, 500, 1, 100, 100, 
                50, 1, 50, 100, 100, 
                50, 1, 100, 1, 1, 
                1, 50, 50, 50, 50, 
                50, 50, 50, 1, 1, 
                1, 1, 50, 100, 1])
```

```python
Brasil_pd = Brasil * p_d
```

```python
Japao_pd = Japao * p_d
```

### Diff total Geral

```python
np.sum(Brasil_pd)
```

```python
np.sum(Japao_pd)
```

```python
diff_geral = np.sum(Brasil_pd) - np.sum(Japao_pd)
diff_geral
```

### Total Geral

```python
total_geral = np.sum(Brasil_pd) + np.sum(Japao_pd)
display(total_geral)
```

### P(x) Brasil

```python
p_x_brasil = (np.sum(Brasil_pd) / total_geral) 
p_x_brasil_cg = (np.sum(Brasil_cg_pd) / total_geral)
p_x_brasil_ = round((p_x_brasil - p_x_brasil_cg) * 100, 4)
p_x_brasil_
```

### P(x) Japão 

```python
p_x_japao = (np.sum(Japao_pd) / total_geral)
p_x_japao_cg = (np.sum(Japao_cg_pd) / total_geral)
p_x_japao_ = round((p_x_japao - p_x_japao_cg)* 100, 4)
p_x_japao_ 
```

### Diff i%

```python
p_x_brasil_ - p_x_japao_
```

### Resultado
+ 2 (Brasil) x 1 (Japão)
